Matrix computation

```python
a = torch.ones(2, 1)
b = torch.ones(1, 2)
```

```python
print(torch.mm(a, b))
print(torch.matmul(a, b))
print(a @ b)
print(a.mm(b))
print(a.matmul(b))
```

对于高维的tensor， 矩阵乘法作用于最后两个维度

```python
a = torch.ones(1,2,3,4)
b = torch.ones(1,2,4,3)
print(a.matmul(b))
print(torch.matmul(a, b))
```

```python
a = torch.tensor([1,2,3], dtype=torch.float, device=torch.device('cpu'))
print(a)
print(torch.pow(a, 2))
print(a.pow(2))
print(a**2)
print(a.pow_(2))
print(a)
```

```python
print(torch.exp(a))
b = a.exp_()
print(b)
```

```python
print(a)
print(torch.sqrt(a))
print(a.sqrt())
print(a.sqrt_())
print(a)
```

```python
print(torch.log2(a))
print(torch.log10(a))
print(torch.log(a))
print(torch.log_(a))
print(a)
```
