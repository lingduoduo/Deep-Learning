```python
import torch

a = torch.rand((3, 4))
print(a)
```

```python
out = torch.chunk(a, 2, dim=0)
print(out[0], out[0].shape)
print(out[1], out[1].shape)
```

```python
out = torch.chunk(a, 2, dim=1)
print(out[0], out[0].shape)
print(out[1], out[1].shape)
```

```python
out = torch.chunk(a, 2, dim=-1)
print(out[0], out[0].shape)
print(out[1], out[1].shape)
```

```python
a = torch.rand((10, 4))
print(a)
```

```python
out = torch.split(a, 3, dim=0)
print(len(out))
```

```python
for t in out:
    print(t, t.shape)
```

```python
out = torch.split(a, 3, dim=1)
print(len(out))
```

```python
for t in out:
    print(t, t.shape)
```

```python
out = torch.split(a, 3, dim=-1)
```

```python
for t in out:
    print(t, t.shape)
```

```python

out = torch.split(a, [1, 3, 6], dim=0)
for t in out:
    print(t, t.shape)
```
