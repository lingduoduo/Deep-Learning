```python
import torch

a = torch.rand(2, 1)
print(a)
```

```python
b = torch.rand(2, 1)
print(b)
```

```python
### p-norm distance
print(torch.dist(a, b, p = 1))
```

```python
print(torch.dist(a, b, p = 2))
```

```python
print(torch.dist(a, b, p = 3))
```

```python
print(torch.norm(a))
```

```python
print(torch.norm(a, p=3))
```

```python
print(torch.norm(a, p='fro'))
```
