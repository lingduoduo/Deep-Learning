```python
import torch
```

```python
dev = torch.device("mps")
```

```python
def create_torch_tensors(device):
    x = torch.rand((10000, 10000), dtype=torch.float32)
    y = torch.rand((10000, 10000), dtype=torch.float32)
    x = x.to(device)
    y = y.to(device)

    return x, y
```

```python
i = torch.tensor([[0, 1, 2], [0, 1, 2]])
print(i)
v = torch.tensor([1, 2, 3])
print(v)
```

```python
tensor = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

# index single element
element = tensor[0, 0]  # 1
print(element)
```

```python
# slice a row
slice_1 = tensor[0, :]  # [1, 2, 3]
print(slice_1)
slice_2 = tensor[:, 1]  # [2, 5]
print(slice_2)
```

```python
# change element
tensor[0, 0] = 7
print(tensor)  # [[7, 2, 3], [4, 5, 6]]
```

```python
# shape attribute
shape = tensor.shape  # torch.Size([2, 3])
print(shape)
print(tensor)
```

```python
# unsqueeze(dim) increase the dimension of the tensor by one
expanded_tensor = tensor.unsqueeze(0)  #  new shape = [1, 2, 3]
print(expanded_tensor.shape)
print(expanded_tensor)
```

```python
# reshape(shape) reshpe the tensor to the given shape
reshaped_tensor = tensor.reshape(6)  # new shape = [6]
print(reshaped_tensor.shape)
print(reshaped_tensor)
```

```python
# transpose(dim0, dim1) change two dimentions of the tensor
transposed_tensor = tensor.transpose(0, 1)  # new shape = [3, 2]
print(transposed_tensor)
```
