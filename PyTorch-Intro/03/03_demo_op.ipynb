```python
import torch
```

```python
## add

a = torch.rand(2, 3)
b = torch.rand(2, 3)

print(a)
print(b)
```

```python
print(a + b)
print(a.add(b))
print(torch.add(a, b))
```

```python
print(a)
print(a.add_(b))
print(a)
```

```python
## sub
print("==== sub res ====")
print(a - b)
print(torch.sub(a, b))
```

```python
print(a.sub(b))
print(a.sub_(b))
print(a)
```

```python
## mul
print("===== mul ====")
print(a * b) # element-wise multiplication
print(torch.mul(a, b))
```

```python
print(a.mul(b))
print(a)
print(a.mul_(b))
print(a)
```

```python
tensor = torch.tensor([1, 1, 1, 1])
matrix_product = tensor @ tensor.t()
print(matrix_product)
```

```python
# div
print("=== div ===")
print(a/b) # element-wise division
print(torch.div(a, b))
print(a.div(b))
print(a.div_(b))
print(a)
```

```python
### matmul

a = torch.ones(2, 1)
b = torch.ones(1, 2)
print(a @ b)
print(a.matmul(b))
print(torch.matmul(a, b))
print(torch.mm(a, b))
print(a.mm(b))
```

```python
## high deimensional tensor

a = torch.ones(1, 2, 3, 4)
print(a)
b = torch.ones(1, 2, 4, 3)
print(b)
print(a.matmul(b).shape)
```
