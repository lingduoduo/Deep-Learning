```python
import torch

a = torch.zeros((2, 4))
b = torch.ones((2, 4))
```

```python
out = torch.cat((a,b), dim=0)
print(out)
```

```python
out = torch.cat((a,b),dim=1)
print(out)
```

```python
#torch.stack

print("torch.stack")
a = torch.linspace(1, 6, 6).view(2, 3)
print(a)
print(a.shape)
```

```python
b = torch.linspace(7, 12, 6).view(2, 3)
print(b)
print(b.shape)
```

```python
out = torch.stack((a, b), dim=2)
print(out)
print(out.shape)
```

```python
print(out[:, :, 0])
print(out[:, :, 1])
```

```python
out = torch.stack((a, b), dim=0)
print(out)
print(out.shape)
```

```python
out = torch.stack((a, b), dim=1)
print(out)
print(out.shape)
```

```python
out = torch.stack((a, b), dim=-1)
print(out)
print(out.shape)
```
