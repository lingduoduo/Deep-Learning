```python
import torch
import numpy as np

a = torch.rand(2, 3)
b = torch.rand(2, 3)

print(a)
print(b)
```

```python
print(torch.eq(a, b))
print(torch.equal(a, b))
```

```python
print(torch.ge(a, b))
print(torch.gt(a, b))
print(torch.le(a, b))
print(torch.lt(a, b))
print(torch.ne(a, b))
```

```python
# return a tensor with all elements set to False
# greater_than = tensor > tensor

a = torch.tensor([[1, 4, 4, 3, 5],
                  [2, 3, 1, 3, 5]])
print(a.shape)
```

```python
print(torch.sort(a, dim=1, descending=False))
```

```python
val, ind = torch.sort(a, dim=1, descending=False)
```

```python
print(val)
```

```python
print(ind)
```

```python
torch.sort(a, dim=-1, descending=False)
```

```python
torch.sort(a, dim=0, descending=False)
```

```python
##topk
a = torch.tensor([[2, 4, 3, 1, 5],
                  [2, 3, 5, 1, 4]])
print(a.shape)

print(torch.topk(a, k=2, dim=1, largest=False))
```

```python
print(torch.kthvalue(a, k=2, dim=0))
```

```python
print(torch.kthvalue(a, k=2, dim=1))
```

```python
print(torch.topk(a, k=3, dim=-1, largest=True))
```
