```python
import torch
import numpy as np
```

```python
# torch.tensor() is a class constructor that creates a tensor from a list or a NumPy array.
# The tensor data type is inferred from the input data type.
list_data = [1, 2, 3, 4]
tensor_from_list = torch.tensor(list_data)
print(tensor_from_list) # tensor([1, 2, 3, 4])
```

```python
# torch.tensor() can also create a tensor from a NumPy array.
np_array = np.array([1, 2, 3])
tensor_from_np_tensor = torch.tensor(np_array)
print(tensor_from_np_tensor) # tensor([1, 2, 3], dtype=torch.int32)
```

```python
torch.tensor([1, 2, 3], dtype=torch.int32)
```

```python
a = torch.Tensor([[1, 2],[3, 4]])
print(a)
print(a.type)
np.array([[1, 2],[3, 4]])
```

```python
b = torch.Tensor(2, 2)
print(b)
```

```python
d = torch.tensor(((1, 2), (3, 4)))
print(d.type())
print(d.type_as(a))
```

```python
d = torch.empty(2,3)
print(d.type())
print(d.type_as(a))
```

```python
# torch.zeros(shape) create tensors with all elements set to 0 
d = torch.zeros(2,3)
print(d.type())
print(d.type_as(a))
```

```python
# torch.ones(shape) create tensors with all elements set to 1
d = torch.ones(2,3)
print(d.type())
print(d.type_as(a))
```

```python
d = torch.zeros_like(d)
print(d.type())
print(d.type_as(a))
```

```python
d = torch.ones_like(d)
print(d.type())
print(d.type_as(a))
```

```python
# torch.rand(shape)：create a tensor with elements randomly sampled from a uniform distribution over [0, 1).
d = torch.rand(2, 3)
print(d.type())
print(d.type_as(a))
```

```python
d = torch.linspace(10, 2, 3)
print(d.type())
print(d.type_as(a))
```

```python
dd = torch.normal(mean=0, std=1, size=(2, 3), out=b)
print(b)
print(dd)
```

```python
d = torch.Tensor(2, 2).uniform_(-1, 1)
print(d.type())
print(d.type_as(a))
```

```python
d = torch.randperm(10)
print(d.type())
print(d.type_as(a))
```

```python
# torch.arange(start, end=None, step=1, dtype=None, layout=torch.strided, device=None, requires_grad=False)：create a 1-D tensor of size end-start/step.
d = torch.arange(2, 10, 2)
print(d.type())
print(d.type_as(a))
```
