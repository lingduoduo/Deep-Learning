```python
import torch

a = torch.rand(2, 3)

print(a)
```

```python
out = torch.reshape(a, (3, 2))

print(out)
```

```python
print(torch.t(out))
```

```python
print(torch.transpose(out, 0, 1))
```

```python
a = torch.rand(1, 2, 3)
print(a)
```

```python
out = torch.transpose(a, 0, 1)
print(out)
print(out.shape)
```

```python
out = torch.squeeze(a)
print(out)
print(out.shape)
```

```python
out = torch.unsqueeze(a, -1)
print(out.shape)
```

```python
out = torch.unbind(a, dim=0)
print(out)
```

```python
out = torch.unbind(a, dim=1)
print(out)
```

```python
out = torch.unbind(a, dim=-1)
print(out)
```

```python
print(a)
print(a.shape)
```

```python
print(torch.flip(a, dims=[2, 1]))
```

```python
out = torch.rot90(a, -1, dims=[0, 2])
print(out)
print(out.shape)
```

```python
out = torch.rot90(a, -1)
print(out)
print(out.shape)
```
