```python
import torch
import torch.nn as nn
```

```python
def loss_func(x, logits):
    '''
    x: [batch_size, seq_len]
    logits: [batch_size, seq_len, vocab_size]
    '''
    shift_labels = x[:,1:]

    shift_logits = logits[:,:-1,:]
    vocab_size = logits.size()[-1]
    
    shift_logits = shift_logits.contiguous().view(-1, vocab_size) # [batch_size * (seq_len-1), vocab_size]
    shift_labels = shift_labels.contiguous().view(-1) # [batch_size * (seq_len -1)]
    
    loss_fn = nn.CrossEntropyLoss(ignore_index=-100)
    loss = loss_fn(shift_logits, shift_labels)
    
    return loss
```

```python
batch_size, seq_len, vocab_size = 32, 20, 100000

x = torch.randint(0, vocab_size, size=(batch_size, seq_len))
logits = torch.rand(size=(batch_size, seq_len, vocab_size))
logits = torch.softmax(logits, dim=-1)

l = loss_func(x, logits)
```
