```python

import torch
import math
import torch.nn as nn
# 这里初步学习一下pytorch
# pytorch里面有network的概念，即nn，forward函数的调用时机
class SinusoidalPositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=4096):
        super().__init__()
        self.pe = torch.zeros(max_len, d_model)  # 初始化位置编码矩阵
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)  # 位置索引 [max_len, 1]
        # pos是奇数 cos(pos/10000**(2i/d_model)) pos是偶数 sin(pos/10000**(2i/d_model)) 
        
        ## 计算三角函数里面的值1/10000**(2i/d_model)
        ## exp(x*(-log(10000)/d_model)) = exp(-log(10000)*x/d_model) = 1/(10000*exp(x/d_model))
        # div_term = torch.exp(
        #     torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model)
        # )  # 频率因子 [d_model/2]
        div_term = 10000.0**(torch.arange(0, d_model, 2).float() / d_model)
        self.pe[:, 0::2] = torch.sin(position * div_term)  # 偶数维度用sin
        self.pe[:, 1::2] = torch.cos(position * div_term)  # 奇数维度用cos
    def forward(self, x):
        return x + self.pe[:x.size(1)]  # 截取与输入序列长度匹配的位置编码并相加
```
