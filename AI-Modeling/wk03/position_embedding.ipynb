```python
import torch
import torch.nn as nn
import math
# pytorch有network的概念，forward函数的调用时机

#基类
class sinuPositionEmbedding(nn.Module):
    def __init__(self, d_model, max_length):
        super().__init__()
        self.pe = torch.zeros(max_length, d_model) #初始化位置编码矩阵
        pos = torch.arange(0, max_length, dtype=torch.float).unsqueeze(1) #[max_length, 1]
        # i是奇数的时候， cos(pos/10000**(2*i/d_model))
        # i是偶数的时候， sin(pos/10000**(2*i/d_model))
        div_term = 10000.0**(torch.arange(0, d_model, 2).float() / d_model)
       #[d_model/2]  numpy pytorch里的 broad cast机制
        self.pe[:, 0::2] = torch.sin(pos/div_term) #[max_length,d_model/2]
        self.pe[:, 1::2] = torch.cos(pos/div_term)
        # [max_length, d_model]
        
    def forward(self, x):
        return x + self.pe[:x.size(0)]
```
