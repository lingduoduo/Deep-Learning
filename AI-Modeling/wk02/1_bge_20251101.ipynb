```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('BAAI/bge-base-zh')
```

```python
sentences_1 = [
    "At the global press conference for 'Singer 2025', Chen Chusheng repeatedly emphasized one point: 'Singing competitions must insist on being live!'",
    "Chen Chusheng spoke bluntly that back when he debuted (in 2007's 'Super Boy'), live broadcasting was an everyday thing. A singing show without a live broadcast has no soul."
]

sentences_2 = [
    "The margin financing balance refers to the difference between the amount of money used to buy stocks on margin and the amount repaid.",
    "The short selling balance refers to the difference between the daily amount of short-sold securities and the amount repaid."
]

embeddings_1 = model.encode(sentences_1)
embeddings_2 = model.encode(sentences_2)
```

```python
embeddings_1
```

```python
similarity = embeddings_1@embeddings_2.T
```

```python
queries = ["What is margin financing balance?", "What is short selling balance?"]
passages = [
    "The margin financing balance refers to the difference between the amount of money used to buy stocks on margin and the amount repaid.",
    "The short selling balance refers to the difference between the daily amount of short-sold securities and the amount repaid."
]

instructions = "Generate a representation for this sentence to retrieve relevant documents: "

q_embeddings = model.encode([instructions + q for q in queries], normalize_embeddings=True)
p_embeddings = model.encode(passages, normalize_embeddings=True)
```

```python
sim = q_embeddings@p_embeddings.T
sim
```
