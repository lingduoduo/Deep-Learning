```python
#模型使用
from staticvectors import StaticVectors

model = StaticVectors("neuml/glove-6B/model.sqlite")
```

```python
vec = model.embeddings(["word"])[0]
```

```python
# 计算两个词向量的相似度
import numpy as np

# 示例词向量
vec_a = model.embeddings(["girl"])[0]
vec_b = model.embeddings(["woman"])[0]

# 余弦相似度
cos_sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))
print("余弦相似度:", cos_sim)  # 输出1.0

# 欧氏距离
euclidean_dist = np.linalg.norm(vec_a - vec_b)
print("欧氏距离:", euclidean_dist)  # 输出5.196
```

```python
#可视化
import numpy as np
from sklearn.manifold import TSNE

word_list = ["boy", "girl", "woman", "man", "king", "queen", "water",
             "apple", "banana", "orange", "pear", "strawberry", "flower",
             
             ]

vectors =model.embeddings(word_list)

tsne = TSNE(n_components=2, random_state=42, perplexity=10)
vectors_2d = tsne.fit_transform(vectors)

```

```python
# 步骤3：可视化
import matplotlib.pyplot as plt
plt.figure(figsize=(10,8))
plt.scatter(vectors_2d[:,0], vectors_2d[:,1])
for i, word in enumerate(word_list):
    plt.annotate(word, (vectors_2d[i,0], vectors_2d[i,1]), fontsize=8)
plt.title("词向量可视化（t-SNE降维）")
plt.show()
```
