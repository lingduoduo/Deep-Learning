```python
#计算两个词的相似度
import numpy as np

#获取emebedding
vecs = model.embeddings(["girl", "woman"])
```

```python
vecs
```

```python
# 余弦相似度的计算
# a*b/|a||b|
cos_sim = np.dot(vecs[0], vecs[1])/ (np.linalg.norm(vecs[0])*np.linalg.norm(vecs[1]))

cos_sim
```

```python
#欧式距离
#|a-b|

euc_dist = np.linalg.norm(vecs[0]-vecs[1])
euc_dist
```

```python
#可视化
import numpy as np
from sklearn.manifold import TSNE

word_list = ["boy", "girl", "woman", "man", "king", "queen", "water",
             "apple", "banana", "orange", "pear", "strawberry", "flower"
             ]
vecs = model.embeddings(word_list)

tsne = TSNE(n_components=2, random_state=42, perplexity=10)
vecs_2d = tsne.fit_transform(vecs)
vecs_2d
```

```python
# 画图可视化
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))
plt.scatter(vecs_2d[:,0], vecs_2d[:, 1])

for i, word in enumerate(word_list):
    plt.annotate(word, (vecs_2d[i,0], vecs_2d[i,1]), fontsize=8)
plt.title("词向量可视化（TSNE降维）")
plt.show()
```
