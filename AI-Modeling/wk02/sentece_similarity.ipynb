```python
#模型下载
from modelscope import snapshot_download
model_dir = snapshot_download('BAAI/bge-base-zh', cache_dir="./")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("BAAI/bge-base-zh")
```

```python
sentences_1 = ["《歌手2025》的全球发布会上，陈楚生反复强调一个观点：“歌唱节目，就得坚持直播！”",
               "陈楚生直言不讳，直播在他出道那会儿（2007年《快乐男声》），本就是家常便饭。不直播的歌唱节目，是没有灵魂的。"]
sentences_2 = ["融资余额是指融资买入股票的金额与融资偿还的金额之间的差额。", "融券余额是指每日卖出的融券金额与偿还的融券金额之间的差额。"]

embeddings_1 = model.encode(sentences_1)
embeddings_2 = model.encode(sentences_2)
```

```python
embeddings_1
```

```python
similarity = embeddings_1@embeddings_2.T
```

```python
embeddings_1[0]@embeddings_2[0].T
```

```python
queries = ['融资余额是什么', '融券余额什么']
passages = ["融资余额是指融资买入股票的金额与融资偿还的金额之间的差额。", "融券余额是指每日卖出的融券金额与偿还的融券金额之间的差额。"]


instructions = "为这个句子生成表示以用于检索相关文章："

new_quries = [instructions+q for q in queries]

embeddings_1 = model.encode(new_quries, normalize_embeddings=True)
embeddings_2 = model.encode(passages, normalize_embeddings=True)

```

```python
embeddings_1@embeddings_2.T
```
