```python
import pandas as pd
```

```python
df = pd.read_parquet("00000.parquet", engine="pyarrow")  # 默认引擎为 auto
print(df.head())
```

```python
import jsonlines

with jsonlines.open("industry_corpus_0_10000.jsonl", "w") as w:
    for i in range(len(df)):
        if i>10000:break
        text = df.iloc[i]["text"]
        w.write(
            {
                "text":text
            }
        )
```
